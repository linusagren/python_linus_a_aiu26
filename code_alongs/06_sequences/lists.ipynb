accessing list

```python
cities = ["göteborg", "varberg", "falkenberg", "stockholm", "malmö", "halmstad"]
print(f"{cities[0] = }")
print(f"{cities[3] = }")
print(f"{cities[-1] = }")
print(f"{cities[-3] = }")

#a strong 'stockholm' is a list of characters
print(f"{cities[-3][0] = }")
print(f"{cities[-3][5] = }")
```

slicing a list

```python
print(f"{cities[1:4]}")
print(f"{cities[3:]}")
```

modify list

```python
#mutate list

cities[3] = "kungsbacka"
cities
```

sort a list

```python
cities.sort()
cities
```

## Loop through list

```python
more_cities = ["stockholm", "örebro", "umeå", "luleå"]

for city in more_cities:
    cities.append(city)

cities
```

create a list of cubes between 0 - 10

```python
cubes = []

for i in range(11):
    cubes.append(i**3)

cubes
```

### List comprehension

```python
cubes = [i**3 for i in range(11)]
cubes
```

```python
even_number_cubed = [i**3 for i in range(11) if not i % 2]
even_number_cubed
```

```python
import matplotlib.pyplot as plt

x = list(range(10))
y = [x**2 + 3 for x in x]

plt.plot(x, y, 'o--')
plt.title(r"$y = x^2 + 3$")
plt.xlabel("x")
plt.ylabel("y")
plt.savefig("nice_graph.png", dpi=250)
```
