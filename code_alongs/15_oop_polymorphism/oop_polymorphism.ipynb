## Polymorphism class method example

```python
class Fish:
    def speak(self):
        print("Blub blub")

    def __str__(self):
        return "Fish swimming around"

class Fox:
    def speak(self):
        NotImplemented

    def __str__(self):
        return "Fox: No one knows how I sound"

fish = Fish()
print(Fish)
print(fish)
```

```python
animals = (Fox(), Fox(), Fish(), Fish())
for animal in animals:
    print(animal)
    animal.speak()
```

## Geometry

```python
import numpy as np

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return self.radius**2*np.pi

    def perimeter(self):
        return 2 * np.pi * self.radius

class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side**2

shapes = [Circle(1), Square(2)]

total_area = 0

for shape in shapes:
    total_area += shape.area()

total_area
```

```python
class Rectangle:
    def __init__(self, side, height):
        self.side = side
        self.height = height


    def area(self):
        return self.side * self.height

    def perimeter(self):
        return (self.side * 2) + (self.height * 2)

    def __repr__(self):
        return f"Rectangle({self.side}, {self.height})"

    def __str__(self):
        return f"The rectangle's side is {self.side} and height is {self.height}."
    
shapes = [Rectangle(3, 5), Rectangle(1, 5), Circle(1), Rectangle(2, 2)]

total_area = 0
total_perimeter = 0

for shape in shapes:
    total_area += shape.area()
    total_perimeter += shape.perimeter()

print(total_area)
print(total_perimeter)
```
