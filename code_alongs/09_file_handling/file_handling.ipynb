## Clean up quotes

```python
import re

i = 1

with open("data/quotes.txt") as f_read, open("data/quotes_clean.txt", "w") as f_write:
    f_write.write("Cool quotes\n\n")
    
    for line in f_read:
        line = line.strip(" \n")

        # Substitute 1 or more spaces with 1 space
        line = re.sub(" +", " ", line)

        if line != "":
            f_write.write(f"{i}. {line}\n")
            i += 1
```
