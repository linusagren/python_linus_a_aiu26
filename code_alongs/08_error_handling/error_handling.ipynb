## Try / Except

- Try-block -> Test dangerous code/input
- Except-block -> Handles error instead of crashing
- Raise -> Throw exception

```python
from numbers import Number

age = input("What's your age? ")

try:
    if not isinstance(age, (Number)):
        raise TypeError(f"Age must be a number, not {type(age)}")
    if not 0 <= age < 125:
        raise ValueError(f"Age must be between 0 and 124, not {age}")
    print(f"You are {age:.0f} years old")
except ValueError as err:
    print(err)
```

```python
while True:
    age = input("What's your age? ")

    try:
        if not age.isdigit():
            raise TypeError(f"Age must be a number, not {type(age)}")

        if not 0 <= age < 125:
            raise ValueError(f"Age must be between 0 and 124, not {age}")
        
        print(f"You are {age:.0f} years old")
        break
    except (TypeError, ValueError) as err:
        print(err)
```
