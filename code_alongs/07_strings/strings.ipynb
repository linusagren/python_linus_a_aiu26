# Strings

## concatenate strings

```python
firstname = "bob"
lastname = "dylan"

print(firstname + " " + lastname)
print(f"{firstname} {lastname}")

greet = f"""
Hello
{firstname.capitalize()} {lastname.capitalize()}
"""

print(greet)
```

## Concatenate using loop

```python
#Meny

#mån: pancake
#tis: pokebowl
#ons: fish n chips
#tor: potatisbullar
#fre: pytt i panna

bamba = ["pancake", "pokebowl", "fish n chips", "potatisbullar", "pytt i panna"]
days = ["mån", "tis", "ons", "tor", "fre"]

print("Meny")
print()

for day, food in zip(days, bamba):
    print(f"{day}: {food}")
```

## Clean strings

```python
subjects = "  math, bIOLogy , phySICs    , programming  "
print(subjects)
print(subjects.strip())

subjects = subjects.strip().split(",")
print(subjects)
```
