# Refactor To Follow SOLID

```python
from abc import ABC, abstractmethod
import json

class Employee:
    def __init__(self, name, phone, salary, tax_rate):
        self.name = name
        self.phone = phone
        self.salary = salary
        self.tax_rate = tax_rate

    @property
    def net_salary(self):
        return self.salary * (1 - self.tax_rate)

class Report(ABC):
    @abstractmethod
    def content(self, employee: Employee) -> str: ...

class TxtReport(Report):
    def content(self, employee: Employee) -> str:
        return (
            "Employee report\n"
            f"Name: {employee.name}\n"
            f"Phone: {employee.phone}\n"
            f"Salary: {employee.salary}\n"
            f"Net Salary: {employee.net_salary}"
        )

class JsonReport(Report):
    def content(self, employee: Employee) -> str:
        return json.dumps({
            "name": employee.name,
            "phone": employee.phone,
            "salary": employee.salary,
            "net_salary": employee.net_salary
        }, indent=3)

def export(text: str, path: str):
    with open(path, "w") as file:
        file.write(text)

e1 = Employee("Anna", "0738271862", 25000, .25)
e1_report = TxtReport().content(e1)
e1_json = JsonReport().content(e1)
export(e1_report, "reports/anna.txt")
export(e1_json, "reports/anna.json")
```
