# if-statements

```python
age = 15
age_limit = 18

if age >= age_limit:
    print("Welcome to the bar.")
else:
    print("You are too young.")
```

```python
# Balder
# Requirement: 140cm & < 190cm
# At least 12 years old

height = 188
age = 14

if height < 190 or height > 140 and age >= 12:
    print("You may ride.")
else:
    print("You may not ride.")
```

if-elif-else

```python
# Exam results
# A: 90
# B: 80
# C: 70
# D: 60
# E: 50
# F: <50
# 
# Let user input a score

try:
    score = int(input("Enter a score: "))

    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    elif score >= 70:
        print("C")
    elif score >= 60:
        print("D")
    elif score >= 50:
        print("E")
    else:
        print("F")
except ValueError:
    print("Score must be a valid whole number")
```

```python
# Is my integer even or odd

number = 42
if number % 2 == 0:
    print("Even.")
else:
    print("Odd.")
```

Nested if-statements

```python
# Check if number is divisible by 5
# If yes, check if it's odd/even

number = 45
if number % 5 == 0:
    if number % 2 == 0:
        print("Even")
    elif number % 2 == 1:
        print("Odd")
else:
    print("Not divisible by 5")
```
