# While statement

- loops when condition True
- "if-loop"

```python
i = 0

while i < 5:
    print(".")
    i += 1 # increments i in each iteration
```

```python
i = 0

while i < 10:
    # **2 to the power of 2
    square = i**2
    print(square, end=" ") # end=" " replaces newline with spaces
    i += 1
```

While statement is good when you are unsure how many iterations you need to do

```python
# Oil leakage
year = 0
total_birds = 80000
minimum_birds = total_birds / 10

while total_birds >= minimum_birds:
    year += 1
    total_birds = total_birds / 2
    print(f"Total birds left after {year:.0f} years: {total_birds:.0f}")
```

# Break

```python
import random as rnd

while True:
    dice = rnd.randint(1, 6)

    guess = int(input("Guess a number (1-6): "))

    if guess == dice:
        print("Winner!")
    else:
        print("Leu-heuu. Suhuuur!")

    play_again = input("Play again? (Y/N) ").capitalize()

    if play_again != "Y":
        print("Goodbye.")
        break
```
