## random

count number of fives in 1,000,000 dice rolls -> calculate proportion of fives

```python
import random as rnd

number_rolls = 1_000_000
count_fives = 0

for roll in range(number_rolls):
    dice = rnd.randint(1, 6)

    if dice == 5:
        count_fives += 1

print(f"{(count_fives / number_rolls)}")
```

## Loop through lists

```python
animals = ["ape", "rabbit", "dog", "fish", "cat"]

for i, animal in enumerate(animals):
    print(f"{i}. {animal}")
```

## Nested loop

```python
for row in range(3):
    for col in range(3):
        print(f"({row},{col})", end=" ")
    print()
```
