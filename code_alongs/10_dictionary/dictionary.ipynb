# Dictionary

```python
person = {
    "name": "Kokchun",
    "age": 35,
    "work": "teacher",
    "interests": ["math", "python", "linux"],
    "phone": 112
}

person
```

```python
print(f"{person["name"] = }")
print(f"{person["age"] = }")
print(f"{person["interests"][1] = }")
```

```python
for key in person:
    print(f"{key}: {person[key]}")
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```

```python
# F<10, E:10, D:20, C: 30, B:40, A:50
grade_limits = {}

for i, char_ in enumerate("ABCDEF"[::-1]):
    grade_limits[char_] = i * 10

grade_limits
```

```python
print("Betygsgränser")
for key, value in reversed(grade_limits.items()):
    print(f"{key}: {value}")
```
