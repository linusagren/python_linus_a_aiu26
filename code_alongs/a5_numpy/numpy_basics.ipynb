# Numpy

start with lists
1. create a list of numbers from -10 -> 10
2. take this list multiply each number with 2

```python
numbers = [number*2 for number in range(-10, 11)]
numbers
```

Multiplication on a list -> multiplies the list and not each element

```python
numbers = list(range(-10,11))
print(numbers*2)
```

## Elementwise operations

- multiply
- power
- subtraction
- addition
- division

```python
import numpy as np

number_array = np.array(numbers)
number_array
```

```python
number_array*2
```

```python
number_array**2
```

```python
number_array-100
```

## Array operations

```python
v = np.array([1,2,5,-3,-2.5])
v
```

```python
print(f"{v[0] = }")
print(f"{v[2:] = }")
print(f"{v[-1:] = }")
print(f"{v[::-1] = }")
```

## Draw a graph

```python
import matplotlib.pyplot as plt
x = range(-3, 4)
y = [-x**2 for x in x]

plt.plot(x, y)
```

```python
x = np.linspace(-3, 3)
x
```

```python
x.shape
```

```python
y = -x**2

plt.plot(x, y)
```
