Simulate 10 dice rolls and append the rolls to a list

```python
import random as rnd

rolls = []

i = 0

while i < 10:
    rolls.append(rnd.randint(1, 6))
    i += 1

print(rolls)
```

Do this once again but now use list comprehension instead to create the list.

```python
rolls = [rnd.randint(1, 6) for _ in range(10)]
print(rolls)
```

a) sort the list in ascending order (*)

```python
rolls.sort()
print(rolls)
```

b) sort the list in descending order (*)

```python
rolls.sort(reverse=True)
print(rolls)
```

c) find the maximum and minimum value in the list (*)

```python
max_ = max(rolls)
min_ = min(rolls)

print(max_)
print(min_)
```

d) calculate the mean (average) of the list (*)

```python
sum_ = sum(rolls)
average = sum_ / len(rolls)

average
```

e) find the median of the list (*)

```python
rolls_length = len(rolls)

median = 0

if rolls_length % 2 == 0:
    median_1 = rolls[(rolls_length // 2) - 1]
    median_2 = rolls[(rolls_length // 2)]

    median = (median_1 + median_2) / 2
else:
    median = rolls[rolls_length // 2]

median
```

f) extract only the unique values from the list. The output should be a list (*)

```python
unique_values = []

for roll in rolls:
    if roll not in unique_values:
        unique_values.append(roll)

unique_values
```
