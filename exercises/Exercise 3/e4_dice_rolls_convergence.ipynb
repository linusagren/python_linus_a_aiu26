## 4. Dice rolls convergence (\*\*)

Simulate:

a) 100 dice rolls and count the number of outcome six. (\*)

```python
import random as rnd
sixes = 0

for i in range(100):
    roll = rnd.randint(1, 6)
    
    if roll == 6:
        sixes += 1

sixes
```

b) 10, 100, 1000, 10000, 100000, 1000000 dice rolls. Count the number of outcome six in each simulation and store it in a list. Compute the probability of outcome six in each simulation. (\*\*)

```python
ROLLS_AMOUNT = [10, 100, 1_000, 10_000, 100_000, 1_000_000, 10_000_000]
sixes = 0
sixes_probability = []

for rolls in ROLLS_AMOUNT:
    for roll in range(rolls):
        current_roll = rnd.randint(1, 6)

        if current_roll == 6:
            sixes += 1

    sixes_probability.append(sixes / rolls)

    print(f"Sixes from {format(rolls, ",")} rolls: {format(sixes, ",")}")
    print(f"Probability of rolling six in {format(rolls, ",")} rolls: {sixes / rolls * 100:.2f}%\n")

    sixes = 0
type(sixes_probability[0])
```

c) Use matplotlib to plot this list. (\*)

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1)

exponents = [f"$10^{x}$" for x in range(1, 8)]

ax.plot(range(7), sixes_probability, '*--')
ax.set_xticks(range(7), exponents)
```
