Create a class called Frac to represent mathematical fractions. The class is instantiated with two instance variables: nominator and denominator. Objects instantiated from this class should have methods for addition, subtraction, multiplication, division using the operators +,-,\*,/. Note that these implemented methods must be mathematically correct. Also implement the following methods:

```python

simplify(self, value = None) # simplifies to most simple form unless value is given

__str__(self) # represent the fraction in a neat way for printing

mixed(self) # represent the fraction in mixed terms

__eq__(self, other) # checks equality by overloading ==

```

Also remember to handle errors and validations.

Example of tests that it should handled:

- 1/2 + 1/3 = 5/6
- 1/2 - 1/3 = 1/6
- 7/6 --> 1 1/6 (mixed)
- 3\*1/2 = 3/2
- 1/2 \* 3 = 3/2
- 1/4 + 2 = 9/4
- 1/4 / 1/2 = 1/2
- 2/4 == 1/2 --> True
- 3/4 += 2 = 11/4

Add unit testing to this class

````python
class Frac:
    def __init__(self, nominator: int, denominator: int):
        self.nominator = nominator
        self.denominator = denominator

    def common_denominator(self, other):
        a, b, c, d = self.nominator, self.denominator, other.nominator, other.denominator
        return a * d, b * c, b * d

    def simplify(self, value = None): # simplifies to most simple form unless value is given
        i = 2

        if value is not None: target = Frac(value.nominator, value.denominator)
        else: target = Frac(self.nominator, self.denominator)

        while i <= max(target.nominator, target.denominator):
            if not target.nominator % i and not target.denominator % i:
                target.nominator = target.nominator // i
                target.denominator = target.denominator // i
                i = 2
            else:
                i += 1

        if target.nominator == target.denominator:
            return Frac(1, 1)
        return target

    def mixed(self):
        if self.simplify() == Frac(1, 1): return "1"

        whole_numbers = 0
        while self.nominator >= self.denominator:
            whole_numbers += 1
            self.nominator = self.nominator - self.denominator

        if self.nominator == 0:
            return str(whole_numbers)

        if whole_numbers > 0:
            return f"{whole_numbers} {self.nominator:.0f}/{self.denominator:.0f}"
        return f"{self.nominator:.0f}/{self.denominator:.0f}"

    def __eq__(self, other):
        simplified_self = self.simplify()
        simplified_other = other.simplify()
        return simplified_self.nominator == simplified_other.nominator and simplified_self.denominator == simplified_other.denominator

    def __add__(self, other):
        if not isinstance(other, Frac | int):
            return NotImplemented

        if isinstance(other, int):
            other = Frac(other, 1)

        common_terms = self.common_denominator(other)
        return self.simplify(Frac(common_terms[0] + common_terms[1], common_terms[2]))

    def __radd__(self, other):
        return self.__add__(other)

    def __repr__(self):
        return f"Frac({self.nominator}, {self.denominator})"

f1 = Frac(5, 8)
f2 = Frac(5, 18)
frac = 3 + f1
print(frac)
````
