````python
class Frac:
    def __init__(self, nominator: int, denominator: int):
        self.nominator = nominator
        self.denominator = denominator

    def simplify(self, value = None) -> str: # simplifies to most simple form unless value is given
        i = 2

        while i <= max(self.nominator, self.denominator):
            if not self.nominator % i and not self.denominator % i:
                self.nominator = self.nominator / i
                self.denominator = self.denominator / i
                i = 2
            else:
                i += 1

        if self.nominator == self.denominator:
            return "1"

        _whole_numbers = 0
        while self.nominator >= self.denominator:
            _whole_numbers += 1
            self.nominator = self.nominator - self.denominator

        if self.nominator == 0:
            return f"{_whole_numbers}"

        if _whole_numbers > 0:
            return f"{_whole_numbers} {self.nominator:.0f}/{self.denominator:.0f}"
        return f"{self.nominator:.0f}/{self.denominator:.0f}"

f = Frac(583, 502)
f.simplify()
````
