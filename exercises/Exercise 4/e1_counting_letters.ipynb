## 1. Counting letters (\*)

Let the user input a word:

a) print out the number of letters in the word. (\*)

```python
word = input("Enter a word: ")
count = len(word)
count
```

b) print out the number of **uppercase** and **lowercase** letters of the word (\*)

```python
import re

uppercase_letters = 0
lowercase_letters = 0

pattern = r"[A-Öa-ö]"

letters = re.findall(pattern, word)
print(letters)

for letter in letters:
    if letter.isupper():
        uppercase_letters += 1
    else:
        lowercase_letters += 1

print(f"Word: {word}")
print(f"Uppercase letters no.: {uppercase_letters}")
print(f"Lowercase letters no.: {lowercase_letters}")
```
