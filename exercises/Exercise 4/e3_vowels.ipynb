## 3. Vowels (\*\*)

Count the number of vowels in this sentence: "Pure mathematics is, in its way, the poetry of logical ideas"

```python
sentence = "Pure mathematics is, in its way, the poetry of logical ideas".lower()
vowels = "eiyäöaouå"
total_vowels = 0

for letter in sentence:
    if letter in vowels:
        total_vowels += 1

total_vowels
```
