## 7. Tram (\*)

Kokchun is a **clumpsy** computer user that doesn't take trams too often. Write a program to prompt the user for:

- number of times he/she wants to take tram in one month
- cost for one ticket
- cost for monthly card

The program should calculate if it's worth for him to buy monthly card or not. Make the program user friendly with clear error messages and ask again in case of input errors.

---

```python
def get_user_input(message):
    while True:
        try:
            number = float(input(message))

            if number <= 0:
                raise ValueError("You must enter a valid number above 0.")
            
            return number
        except ValueError as err:
            print(err)

tickets = get_user_input("Enter how many times you take the tram in a month: ")
ticket_price = get_user_input("Price for one ticket: ")
monthly_ticket_price = get_user_input("Price for a monthly ticket: ")

price_diff = (ticket_price * tickets) - monthly_ticket_price

if price_diff > 0:
    print(f"Get a monthly ticket. Single tickets are ~{price_diff:.0f} more expensive.")
elif price_diff == 0:
    print(f"Consider a monthly ticket. You'd spend as much on single tickets anyway.")
else:
    print(f"Stick with single tickets. A monthly card is ~{-(price_diff):.0f} more expensive")
```
